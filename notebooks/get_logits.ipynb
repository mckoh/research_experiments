# Logits from LLM

In diesem Experiment greife ich direkt die Logits Werte eines GPT2 LLMs ab, ohne diese durch die Softmax-Funktion zu schicken.

```python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer

model_name = "gpt2"  # Oder ein bel. Causal-LM
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

inputs = tokenizer("Das Ergebnis lautet", return_tensors="pt")

with torch.no_grad():
    outputs = model(**inputs)
    # outputs.logits hat die Form: [batch_size, sequence_length, vocab_size]
    logits = outputs.logits

# Logits für das allerletzte eingegebene Token (Vorhersagebasis für das nächste Token):
next_token_logits = logits[0, -1, :]

print("Shape der Logits:", next_token_logits.shape)  # Tensor der Größe [vocab_size]
print("Erste 10 Logits-Rohwerte:", next_token_logits[:10])
```
