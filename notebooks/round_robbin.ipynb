# Round Robbin

Implements an automated process on how to scan a rectangular field.

```python
from numpy import array


SIZE = 4


def walk(main, cross, x, y, size):

    DIRECTION = {
        "left": {"dx": -1, "dy": 0, "opp": "right"},
        "right": {"dx": 1, "dy": 0, "opp": "left"},
        "up": {"dx": 0, "dy": -1, "opp": "down"},
        "down": {"dx": 0, "dy": 1, "opp": "up"}
    }

    for _ in range(size):
        for i in range(size):

            # Show where drone currently is
            print(f"X: {x:2d}, Y: {y:2d}")

            if i < size - 1:
                if main == "left" or main == "right":
                    x += DIRECTION[main]["dx"]
                elif main == "up" or main == "down":
                    y += DIRECTION[main]["dy"]

        if main == "left" or main == "right":
            y += DIRECTION[cross]["dy"]
        elif main == "up" or main == "down":
            x += DIRECTION[cross]["dx"]
        main = DIRECTION[main]["opp"]

    return main, cross, x, y-1

if __name__ == "__main__":

    x = 0
    y = 0

    last_main, last_cross, x, y = walk("right", "down", x, y, SIZE)
    last_main, last_cross, x, y = walk("up", "right", x, y, SIZE)
```
