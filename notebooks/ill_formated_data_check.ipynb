# Clean CSV Files on Loading

In this file I experiment with a loading function that detects ill-formated rows in a CSV (ones that have less columns than the header) and skips them automatically.

```python
from pandas import read_csv
import csv


DATA_FILE = "../data/ill_formated_data.csv"


def load_data(data_file, sep=",", *args, **kwargs):
    """Checks data for ill-formated lines and loads data """

    ill_lines = []

    with open(data_file, mode='r', encoding='utf-8') as f:
        reader = csv.reader(f, delimiter=sep)

        header_length = len(next(reader))

        for line_num, row in enumerate(reader, start=2): # start=2 wegen Header
            if len(row) != header_length:
                print(f"Achtung in Zeile {line_num}: Hat {len(row)} Spalten, erwartet wurden {header_length}.")
                ill_lines.append(line_num-1)

    return read_csv(
        data_file,
        sep=sep,
        skiprows=ill_lines,
        *args, **kwargs
    )
```

```python
dtype_dict={
    "column_0": "object",
    "column_1": "Int64",
    "column_2": "object",
    "column_3": "Int64"
}

df = load_data(DATA_FILE, sep=",", na_values="?", dtype=dtype_dict)
df.head()
```
