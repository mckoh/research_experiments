# OLLAMA Endless Thinking

Kleines Experiment, bei dem ich ein `llama3.1` Modell in eine lange Gedankenschleife schicke um in 1000 Iterationen auszuprobieren, wohin mich das führt.

```python
import ollama
from IPython.display import clear_output


TEXT_FILE = "../output/responses.txt"
N_ITERATIONS = 1000


def generate_response(prompt_text: str) -> str:

    system_instruction = """
        Du bist ein hochanalytischer Denkpartner und Experte für
        sokratisches Hinterfragen. Deine einzige Aufgabe ist es,
        mit dem Nutzer eine tiefe, mehrstufige Exploration eines
        Themas durchzuführen.

        VERHALTENSREGELN:
        1. Analysiere jede Eingabe des Nutzers auf zugrundeliegende Annahmen, Kernkonzepte, logische Implikationen oder blinde Flecken.
        2. Formuliere als Antwort ausschließlich eine präzise, tiefergehende Folgefrage.
        3. Gib keine Erklärungen, Standardantworten oder Zusammenfassungen. Deine gesamte Ausgabe besteht nur aus dieser Folgefrage.
        4. Jede neue Frage muss logisch auf den vorherigen Schritten aufbauen und den Abstraktions- bzw. Vertiefungsgrad erhöhen (z. B. von der Anwendung zu den Grundlagen, von Symptomen zu Ursachen, oder von Fakten zu ethischen/systemischen Implikationen).
        5. Halte den Ton analytisch, sachlich und fokussiert.
        6. Formuliere deine Antwort in maximal 200 Worten.

        ZIEL:
        Bringe den Nutzer Schritt für Schritt dazu, das Thema bis zu seinen fundamentalen Prinzipien, Grenzen und Zusammenhängen zu durchdenken.
    """

    response = ollama.chat(
        model="llama3.1",
        messages=[
            {"role": "system", "content": system_instruction},
            {"role": "user", "content": prompt_text},
        ],
    )

    return response["message"]["content"]


def get_last_line(file):

    with open(file, "r", encoding="utf-8") as file:
        lines = file.readlines()
        last_line = lines[-1] if lines else None

    return last_line.strip(), len(lines)


def clean(prompt):

    prompt = prompt.strip()
    prompt = prompt.replace("'", " ")
    prompt = prompt.replace("\n", " ")
    for _ in range(10):
        prompt = prompt.replace("  ", " ")
    return prompt
```

```python
prompt, n = get_last_line(TEXT_FILE)

responses = []

for i in range(N_ITERATIONS):
    clear_output(wait=True)
    prompt = generate_response(prompt)
    prompt = prompt.strip().replace("\"", " ").replace("  ", " ")

    with open(TEXT_FILE, "a", encoding="utf-8") as file:
        file.write(clean(prompt) + "\n")

    print(i+n, "Denkschritte bereits erledigt. Ich mache noch", N_ITERATIONS-i-1, "weitere Schritte.")
```
