```python
import cv2
import os

def extract_frames(video_path, output_folder, interval_seconds=10):
    """
    Extrahiert Frames aus einem MP4-Video in einem festen Sekunden-Intervall
    und speichert sie als PNG-Bilder ab.
    """
    # Zielordner erstellen, falls er noch nicht existiert
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
        print(f"Zielordner '{output_folder}' wurde erfolgreich erstellt.")

    # Video öffnen
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        print(f"Fehler: Video '{video_path}' konnte nicht geöffnet werden.")
        return

    # Framerate (FPS) des Videos auslesen
    fps = cap.get(cv2.CAP_PROP_FPS)
    if fps == 0:
        print("Fehler: Die Framerate des Videos konnte nicht ermittelt werden.")
        return

    # Berechnen, nach wie vielen Frames die 10 Sekunden erreicht sind
    frame_interval = int(fps * interval_seconds)
    print(f"Video-FPS: {fps:.2f} | Exportiere alle {interval_seconds} Sekunden (alle {frame_interval} Frames).")

    frame_count = 0
    saved_count = 0

    while True:
        ret, frame = cap.read()

        # Abbrechen, wenn das Video zu Ende ist
        if not ret:
            break

        # Prüfen, ob das Intervall erreicht ist
        if frame_count % frame_interval == 0:
            # Aktuelle Videozeit berechnen für einen sprechenden Dateinamen
            total_seconds = int(frame_count / fps)
            minutes = total_seconds // 60
            seconds = total_seconds % 60

            # Dateiname erzeugen (z.B. frame_00_10.png, frame_00_20.png)
            filename = os.path.join(output_folder, f"frame_{minutes:02d}_{seconds:02d}.png")

            # Frame als PNG speichern
            cv2.imwrite(filename, frame)
            print(f"Gespeichert: {filename}")
            saved_count += 1

        frame_count += 1

    # Ressourcen wieder freigeben
    cap.release()
    print(f"\nExtraktion abgeschlossen! Insgesamt {saved_count} Bilder in '{output_folder}' gesichert.")
```

```python
# --- BEISPIEL FÜR DEN AUFRUF ---
# Ersetzen Sie 'wasserbad_video.mp4' mit Ihrem Dateipfad
video_datei = "../data/magna_leckage_pruefung.mp4"
ziel_ordner = "../data/magna_leckage_pruefung"

# Funktion ausführen
extract_frames(video_datei, ziel_ordner, interval_seconds=2)
```
