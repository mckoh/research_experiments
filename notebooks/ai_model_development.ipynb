# Development of AI Models

This notebook analyses, how AI models have developed over the years. The datasets belong to: https://hai.stanford.edu/ai-index/2026-ai-index-report.

```python
from pandas import read_csv, to_datetime
```

```python
data = read_csv("../data/ai-models/frontier_ai_models.csv")
data.fillna({"Country (of organization)": "None"}, inplace=True)

transformer = lambda x: x.split(",")[0] if x else "leer"
data["Country (of organization)"] = data["Country (of organization)"].map(transformer)
data["Publication date"] = to_datetime(data["Publication date"])
data.index = data["Publication date"]
```

```python
data.to_csv("../data/ai-models/final.csv")
```
