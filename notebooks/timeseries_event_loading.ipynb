In dieser Demo habe ich mich mit dem Laden von JSON-Event-Logs beschäftigt. Dieses Verfahren ist die basis für das gemeinsame Projekt mit Liebherr/Prinoth. Wir planen eine Analyse von einschlägigen Maschinen-Logs zur Untersuchung von Mustern innerhalb dieser Daten. Inhalt des Projekts ist zum einen das Finden einer passenden Datenstruktur die für beide Unternehmen passen kann und andererseits einer Lade/Verarbeitungs/Analysetechnik.

```python
from pandas import read_json
```

## Load Data

```python
data = read_json(
    "../data/liebherr_prinoth_demo.json",
    convert_dates=["ts_utc"],
    dtype={
        'machine_id': 'object',
        'ts_utc': "datetime",
        'event_id': "object",
        "group": "object",
        "code": "object",
        "severity": "object",
        "priority": "int64",
        "text": "str",
        "value": "float64",
        "unit": "object",
        "source": "object"
    }
)
```

```python
from pandas import read_csv

class EventLoader:

    def __init__(self, use_columns=False, ts_column=0, dim_columns=False, fact_columns=False, file_path = None):

        self.use_columns = use_columns
        self.ts_column = ts_column
        self.dim_columns = dim_columns
        self.fact_columns = fact_columns
        self.file_path = file_path

        self.data = None

    def read_csv(self):
        data = read_csv(self.file_path)
        return data
```
