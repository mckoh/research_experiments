# Logits to Probabilies

In this experiment I show how logits are adjusted into probabilities using temperature and softmax function. To do this, I use the example of my workshop "Prompt rein - Text raus".

![](../data/logits.png)

```python
from numpy import exp

words = [
    "anmelden",
    "registrieren",
    "abmelden",
    "kochen",
    "bestellen",
    "testen"
]

logits = [2.4, 2.1, -0.8, -3.6, -3.0, -4.2]

adjust_logits = lambda temp: [z/temp for z in logits]
softmax = lambda temp: [exp(s/temp) / sum([exp(z/temp) for z in adjust_logits(temp)]) for s in adjust_logits(temp)]
print_words = lambda words, temp: [(words[i], softmax(temp)[i]) for i in range(len(words))]
```

```python
print_words(words, 0.3)
```

```python
print_words(words, 1)
```

```python
print_words(words, 2)
```
