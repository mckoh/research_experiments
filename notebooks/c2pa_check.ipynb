# C2PA-Check in Python

In diesem Notebook habe ich mit Hilfe des C2PATool (Brew install c2patool) einen entsprechenden Check in Python implementiert, der das C2PA-Manifest eines Bildes ausliest und darstellt.

```python
import json
import subprocess

def check_c2pa(image_path):
    try:
        result = subprocess.run(
            ["c2patool", image_path],
            capture_output=True,
            text=True,
            check=True
        )
        data = json.loads(result.stdout)
        return data
    except subprocess.CalledProcessError:
        return "Kein C2PA-Manifest vorhanden."
    except FileNotFoundError:
        return "c2patool ist nicht im systemweiten PATH installiert."

# Anwendung
manifest = check_c2pa("../data/gemini_image.jpg")
print(manifest)
```
